```python
import time
import mlflow
from sentence_transformers import SentenceTransformer, util

mlflow.set_experiment("embedding-model-selection")

models = {
    "bge": "BAAI/bge-small-en-v1.5",
    "minilm": "sentence-transformers/all-MiniLM-L6-v2",
    "mpnet": "sentence-transformers/paraphrase-mpnet-base-v2"
}

test_data = [
    ("Built ML models using sklearn", "Experience with machine learning and scikit-learn", 1),
    ("Frontend developer with React", "Looking for backend Python engineer", 0),
    ("Worked on NLP projects", "Experience in text processing and NLP", 1),
    ("Android app development", "Data science and ML role", 0),
]

for name, model_path in models.items():
    with mlflow.start_run(run_name=name):

        model = SentenceTransformer(model_path)

        similarities = []
        good_scores = []
        poor_scores = []

        start_time = time.time()

        for s1, s2, label in test_data:
            emb1 = model.encode(s1, convert_to_tensor=True)
            emb2 = model.encode(s2, convert_to_tensor=True)

            score = util.cos_sim(emb1, emb2).item()
            similarities.append(score)

            if label == 1:
                good_scores.append(score)
            else:
                poor_scores.append(score)

        end_time = time.time()

        # Metrics
        avg_good = sum(good_scores) / len(good_scores)
        avg_poor = sum(poor_scores) / len(poor_scores)
        inference_time = (end_time - start_time) * 1000  # ms

        # Logging
        mlflow.log_param("model_name", model_path)
        mlflow.log_metric("avg_cosine_good_match", avg_good)
        mlflow.log_metric("avg_cosine_poor_match", avg_poor)
        mlflow.log_metric("inference_time_ms", inference_time)
        mlflow.log_metric("separation", avg_good - avg_poor)

        print(f"{name} done")
```

```python
import mlflow
import time
from sentence_transformers import SentenceTransformer, util

# Load model
model_name = "sentence-transformers/all-MiniLM-L6-v2"
model = SentenceTransformer(model_name)

mlflow.set_experiment("resume_jd_threshold_tuning")

# Example dataset format:
# dataset = [{"resume": "...", "jd": "...", "label": 1}, ...]

thresholds = [0.4, 0.5, 0.6, 0.7,0.8]
batch_sizes = [2]  # can expand later

for threshold in thresholds:
    for batch_size in batch_sizes:

        with mlflow.start_run(run_name=f"thr_{threshold}_bs_{batch_size}"):

            correct = 0
            total = len(dataset)

            good_scores = []
            poor_scores = []

            TP = FP = TN = FN = 0

            start_time = time.time()

            for sample in dataset:
                emb1 = model.encode(sample["resume"], convert_to_tensor=True)
                emb2 = model.encode(sample["jd"], convert_to_tensor=True)

                score = util.cos_sim(emb1, emb2).item()
                label = sample["label"]

                pred = 1 if score >= threshold else 0

                # Accuracy
                if pred == label:
                    correct += 1

                # Confusion matrix
                if pred == 1 and label == 1:
                    TP += 1
                elif pred == 1 and label == 0:
                    FP += 1
                elif pred == 0 and label == 0:
                    TN += 1
                elif pred == 0 and label == 1:
                    FN += 1

                # Score tracking
                if label == 1:
                    good_scores.append(score)
                else:
                    poor_scores.append(score)

            end_time = time.time()

            # Metrics
            accuracy = correct / total
            avg_good = sum(good_scores) / len(good_scores)
            avg_poor = sum(poor_scores) / len(poor_scores)
            separation = avg_good - avg_poor

            precision = TP / (TP + FP) if (TP + FP) > 0 else 0
            recall = TP / (TP + FN) if (TP + FN) > 0 else 0
            f1 = (2 * precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

            inference_time = (end_time - start_time) * 1000

            # Log to MLflow
            mlflow.log_param("model", model_name)
            mlflow.log_param("threshold", threshold)
            mlflow.log_param("batch_size", batch_size)

            mlflow.log_metric("accuracy", accuracy)
            mlflow.log_metric("precision", precision)
            mlflow.log_metric("recall", recall)
            mlflow.log_metric("f1_score", f1)

            mlflow.log_metric("avg_good_score", avg_good)
            mlflow.log_metric("avg_poor_score", avg_poor)
            mlflow.log_metric("separation", separation)
            mlflow.log_metric("inference_time_ms", inference_time)

            print(f"Done threshold {threshold}")
```

```python
def evaluate(dataset, threshold):
    TP = FP = TN = FN = 0
    good_scores = []
    poor_scores = []

    for sample in dataset:
        emb1 = model.encode(sample["resume"], convert_to_tensor=True)
        emb2 = model.encode(sample["jd"], convert_to_tensor=True)

        score = util.cos_sim(emb1, emb2).item()
        label = sample["label"]

        pred = 1 if score >= threshold else 0

        if pred == 1 and label == 1: TP += 1
        elif pred == 1 and label == 0: FP += 1
        elif pred == 0 and label == 0: TN += 1
        elif pred == 0 and label == 1: FN += 1

        if label == 1:
            good_scores.append(score)
        else:
            poor_scores.append(score)

    precision = TP/(TP+FP) if TP+FP else 0
    recall = TP/(TP+FN) if TP+FN else 0
    f1 = (2*precision*recall)/(precision+recall) if precision+recall else 0
    accuracy = (TP+TN)/(TP+TN+FP+FN)

    avg_good = sum(good_scores)/len(good_scores)
    avg_poor = sum(poor_scores)/len(poor_scores)
    separation = avg_good - avg_poor

    return {
        "precision": precision,
        "recall": recall,
        "f1": f1,
        "accuracy": accuracy,
        "separation": separation
    }
```

```python
best_threshold = 0
best_f1 = 0

for t in [i/100 for i in range(30, 90, 5)]:
    result = evaluate(dataset, t)

    if result["f1"] > best_f1:
        best_f1 = result["f1"]
        best_threshold = t

print("Best Threshold:", best_threshold)
```

```python
import mlflow
import time

mlflow.set_experiment("advanced_resume_jd_system")
```

```python
def run_experiment(dataset, thresholds):

    for threshold in thresholds:

        with mlflow.start_run(run_name=f"threshold_{threshold}"):

            TP = FP = TN = FN = 0
            good_scores = []
            poor_scores = []
            errors = 0

            start_time = time.time()

            for sample in dataset:
                resume = sample["resume"]
                jd = sample["jd"]
                label = sample["label"]

                # embeddings
                emb1 = model.encode(resume, convert_to_tensor=True)
                emb2 = model.encode(jd, convert_to_tensor=True)

                cosine = util.cos_sim(emb1, emb2).item()

                # skill overlap
                overlap = skill_overlap(resume, jd)

                # final score
                score = 0.7 * cosine + 0.3 * overlap

                # calibration
                prob = calibrate(score)

                pred = 1 if score >= threshold else 0

                # confusion matrix
                if pred == 1 and label == 1: TP += 1
                elif pred == 1 and label == 0: FP += 1
                elif pred == 0 and label == 0: TN += 1
                elif pred == 0 and label == 1: FN += 1

                if pred != label:
                    errors += 1

                # score tracking
                if label == 1:
                    good_scores.append(score)
                else:
                    poor_scores.append(score)

            end_time = time.time()

            # metrics
            precision = TP/(TP+FP) if TP+FP else 0
            recall = TP/(TP+FN) if TP+FN else 0
            f1 = (2*precision*recall)/(precision+recall) if precision+recall else 0
            accuracy = (TP+TN)/(TP+TN+FP+FN)

            avg_good = sum(good_scores)/len(good_scores)
            avg_poor = sum(poor_scores)/len(poor_scores)
            separation = avg_good - avg_poor

            inference_time = (end_time - start_time) * 1000

            # -------- LOGGING --------
            mlflow.log_param("threshold", threshold)
            mlflow.log_param("model", "MiniLM + overlap")

            mlflow.log_metric("accuracy", accuracy)
            mlflow.log_metric("precision", precision)
            mlflow.log_metric("recall", recall)
            mlflow.log_metric("f1_score", f1)

            mlflow.log_metric("avg_good_score", avg_good)
            mlflow.log_metric("avg_poor_score", avg_poor)
            mlflow.log_metric("separation", separation)

            mlflow.log_metric("error_count", errors)
            mlflow.log_metric("inference_time_ms", inference_time)

            print(f"Done threshold {threshold}")
```

```python
thresholds = [0.4, 0.5, 0.6,0.65, 0.7,0.72,0.75, 0.8]

run_experiment(dataset, thresholds)
```

```python
important_skills = [
    "python", "machine learning", "aws", "docker",
    "sql", "react", "kubernetes"
]

def extract_skills(text):
    text = text.lower()
    return [skill for skill in important_skills if skill in text]

def skill_gap(resume, jd):
    r = set(extract_skills(resume))
    j = set(extract_skills(jd))

    return list(j - r)
```

```python
def analyze_resume_jd(resume, jd):

    # embeddings
    emb1 = model.encode(resume, convert_to_tensor=True)
    emb2 = model.encode(jd, convert_to_tensor=True)

    # cosine similarity
    cosine = util.cos_sim(emb1, emb2).item()

    # skill overlap
    overlap = skill_overlap(resume, jd)

    # final score
    final_score = 0.7 * cosine + 0.3 * overlap

    # probability
    probability = calibrate(final_score)

    # label
    if final_score >= THRESHOLD:
        match_level = "Strong Match"
    elif final_score >= 0.5:
        match_level = "Moderate Match"
    else:
        match_level = "Low Match"

    # missing skills
    missing_skills = skill_gap(resume, jd)

    return {
        "match_score": round(final_score, 2),
        "match_level": match_level,
        "hiring_probability": f"{round(probability * 100)}%",
        "missing_skills": missing_skills[:10]  # limit output
    }
```
